```python
import pandas as pd
import json
```

```python
boris = json.load(open("B1 Adalimumab (1).boris"))
```

```python
videos = list(boris['observations'].keys())
```

```python
for video in videos:
    events = [x for x in boris['observations'][video]['events'] if 'Scratch' in x]
    df = pd.DataFrame(events)
    for m in df[1].unique():
        df_m = df[df[1] == m]
        df_m.to_csv(video + "_" + str(m) + ".csv", index=False)

```
